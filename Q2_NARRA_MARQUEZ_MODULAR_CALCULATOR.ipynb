**Why is it better to divide the calculator program into functions instead of** **writing everything in one long block of code?**

Using functions instead of one massive block of code is a fundamental programming principle called modularity. It is better for several reasons:
Readability for Cleaner Code It separates what the program does from how it does it. Reusability If you want to perform addition or division in five different places throughout a larger application, you only have to write the math logic once. You can just call add_numbers() over and over again without rewriting the code.


```python
# Mini Calculator Game

# 1. Define the functions first
def is_valid_range(value):
    # Returns True if value is between 1 and 4 (inclusive)
    return 1 <= value <= 4

def add_numbers(num1, num2):
    return num1 + num2

def subtract_numbers(num1, num2):
    return num1 - num2

def multiply_numbers(num1, num2):
    return num1 * num2

def divide_numbers(num1, num2):
    # Prevent division by zero
    if num2 == 0:
        return "Error: Division by zero"
    return num1 / num2

# Ask the user to enter the first number
num1 = float(input("Enter number!! "))

# Ask the user to enter the second number
num2 = float(input("Enter number!! "))

# Display the operations
print("Choose operation: 1 - Addition 2 - Subtraction 3 - Multiplication 4 - Division")

# Loop strictly for the operation choice
while True:
    # Ask the user to enter the operation choice
    value = float(input("Enter Operation!! "))

    # Validation
    if is_valid_range(value):
        print("Input is within the valid range.")
        break # Success! Break out of this small loop and move down to calculations
    else:
        print("Error: Input is out of bounds.")
        continue # Jumps back to "Enter Operation!!" without touching num1 or num2

# 3. Check the choice and call the correct function
if value == 1:
    result = add_numbers(num1, num2)
elif value == 2:
    result = subtract_numbers(num1, num2)
elif value == 3:
    result = multiply_numbers(num1, num2)
elif value == 4:
    result = divide_numbers(num1, num2)
else:
    result = "Invalid choice"

# 4. Print the final result
print("Result:", result)
```
